The first cell loads the four Python *libraries* (toolboxes) we use all term. Part 0 explains what each one does. The second cell is hidden behind a title in Colab. It holds helper code for the auto-checks and a few drawings. **Just run it; you do not need to read it.**

````python
# ⚙️ SETUP CELL: loads our toolboxes and sets up readable, colorblind-safe charts
import numpy as np                  # numbers, square roots, arrays
import pandas as pd                 # data tables ("DataFrames")
import matplotlib.pyplot as plt     # charts
import sympy as sp                  # symbolic algebra: x, y, expand, factor, solve
sp.init_printing()                  # show SymPy results as nicely formatted math
np.random.seed(2026)                # makes anything "random" reproducible

# Okabe–Ito palette: distinguishable for people with color-vision deficiencies
OKABE_ITO = ["#0072B2", "#E69F00", "#009E73", "#D55E00", "#CC79A7", "#56B4E9", "#F0E442", "#000000"]
plt.rcParams.update({"figure.figsize": (8, 4.5), "font.size": 12, "axes.titlesize": 14,
                     "axes.labelsize": 12, "axes.grid": True, "grid.alpha": 0.3,
                     "axes.prop_cycle": plt.cycler(color=OKABE_ITO)})
pd.set_option("display.max_columns", 20)
print("✅ Setup complete. pandas", pd.__version__, "| numpy", np.__version__, "| sympy", sp.__version__)
````

````python
#@title 🔧 Helper code: run me (no need to read) { display-mode: "form" }
# Helper code. You do NOT need to understand this cell; just run it.
REGION_NAMES = {13: "NCR", 14: "CAR", 1: "Ilocos", 2: "Cagayan Valley", 3: "Central Luzon",
                4: "CALABARZON", 17: "MIMAROPA", 5: "Bicol", 6: "Western Visayas",
                7: "Central Visayas", 8: "Eastern Visayas", 9: "Zamboanga Peninsula",
                10: "Northern Mindanao", 11: "Davao", 12: "SOCCSKSARGEN", 15: "ARMM", 16: "Caraga"}
SEX = {1: "Male", 2: "Female"}
MARITAL = {1: "Single", 2: "Married", 3: "Widowed", 4: "Divorced/Separated", 5: "Annulled", 6: "Unknown"}
EMPSTAT = {1: "Employed", 2: "Unemployed", 3: "Not in labor force"}
WORKER_CLASS = {0: "Private household", 1: "Private establishment", 2: "Government",
                3: "Self-employed", 4: "Employer", 5: "Family business (with pay)",
                6: "Family business (without pay)"}
PAY_BASIS = {0: "In kind", 1: "Per piece", 2: "Per hour", 3: "Per day", 4: "Monthly",
             5: "Pakyaw", 6: "Other", 7: "Commission"}
EDUC_LEVELS = ["Elementary or less", "High school", "Post-secondary/technical", "College or higher"]

def educ_level(code):
    """Recode the 5-digit LC07_GRADE code into 4 broad levels (instructor: verify vs. dictionary)."""
    if pd.isna(code):
        return np.nan
    code = int(code)
    if code < 20000:
        return EDUC_LEVELS[0]      # no grade, preschool, elementary
    if code < 40000:
        return EDUC_LEVELS[1]      # junior and senior high school
    if code < 60000:
        return EDUC_LEVELS[2]      # post-secondary non-tertiary, short-cycle/technical
    return EDUC_LEVELS[3]          # bachelor's and above

def check(answer, expected, hint="Re-read the task and try again.", tol=1e-4, exact=False):
    """Compare a student's answer with the expected one and print friendly feedback."""
    if answer is None:
        print("⏳ Not attempted yet. Replace None with your code, run that cell, then run this check again.")
        return
    try:
        if isinstance(expected, sp.Set) or exact:
            assert answer == expected
        elif isinstance(expected, sp.Basic):
            assert sp.simplify(sp.sympify(answer) - expected) == 0
        elif isinstance(expected, (set, frozenset, str, bool, np.bool_)):
            assert answer == expected
        elif isinstance(expected, (list, tuple, np.ndarray, pd.Series)):
            a, e = np.asarray(answer, dtype=float), np.asarray(expected, dtype=float)
            assert a.shape == e.shape and np.allclose(a, e, rtol=tol, atol=1e-6)
        else:
            assert np.isclose(float(answer), float(expected), rtol=tol, atol=1e-6)
        print("✅ Correct! Well done.")
        return
    except Exception:
        print("❌ Not quite yet. Hint: " + hint)
        return

def number_line(intervals, xlim=(0, 100), title="Intervals on the number line", xlabel="Value"):
    """Draw intervals. Each item: (a, b, left_closed, right_closed, label). Use np.inf for ∞."""
    fig, ax = plt.subplots(figsize=(8, 1.2 + 0.7 * len(intervals)))
    for i, (a, b, lc, rc, label) in enumerate(intervals):
        y, color = len(intervals) - i, OKABE_ITO[i % len(OKABE_ITO)]
        lo, hi = max(a, xlim[0]), min(b, xlim[1])
        ax.plot([lo, hi], [y, y], color=color, lw=5, solid_capstyle="butt")
        for x, closed in ((a, lc), (b, rc)):
            if np.isfinite(x):
                ax.plot(x, y, "o", ms=11, color=color, mfc=color if closed else "white", mew=2.5)
        if not np.isfinite(b):
            ax.plot(xlim[1], y, ">", ms=12, color=color)
        if not np.isfinite(a):
            ax.plot(xlim[0], y, "<", ms=12, color=color)
        ax.text(lo, y + 0.22, label, fontsize=11, color=color)
    ax.set_yticks([]); ax.set_xlim(xlim[0] - 2, xlim[1] + 2); ax.set_ylim(0.4, len(intervals) + 0.8)
    ax.set_title(title); ax.set_xlabel(xlabel)
    plt.show()

def venn2_counts(only_a, only_b, both, neither, labels=("A", "B"), title="Venn diagram"):
    """Draw a simple two-set Venn diagram with counts in each region."""
    from matplotlib.patches import Circle, Rectangle
    fig, ax = plt.subplots(figsize=(7, 4.5))
    ax.add_patch(Rectangle((-2.6, -1.9), 5.2, 3.8, fill=False, lw=1.5))
    ax.add_patch(Circle((-0.7, 0), 1.5, alpha=0.35, color=OKABE_ITO[0]))
    ax.add_patch(Circle((0.7, 0), 1.5, alpha=0.35, color=OKABE_ITO[1]))
    for x, y, txt in ((-1.4, 0, only_a), (1.4, 0, only_b), (0, 0, both), (2.0, -1.6, neither)):
        ax.text(x, y, f"{txt:,}", ha="center", va="center", fontsize=13, fontweight="bold")
    ax.text(-1.6, 1.6, labels[0], fontsize=13, ha="center"); ax.text(1.6, 1.6, labels[1], fontsize=13, ha="center")
    ax.text(-2.45, -1.65, "U", fontsize=13)
    ax.set_xlim(-2.8, 2.8); ax.set_ylim(-2.1, 2.1); ax.set_aspect("equal"); ax.axis("off"); ax.set_title(title)
    plt.show()

def make_practice_data(n=3000, seed=2026):
    """PRACTICE ONLY – NOT REAL DATA. Synthetic people with the same columns as the teaching file."""
    rng = np.random.default_rng(seed)
    codes = np.array(list(REGION_NAMES))
    probs = np.where(np.isin(codes, [13, 4, 3]), 3.0, 1.0); probs = probs / probs.sum()
    region = rng.choice(codes, n, p=probs)
    age = np.clip(15 + rng.gamma(2.0, 14.0, n), 15, 95).astype(int)
    sex = rng.choice([1, 2], n)
    educ = rng.choice(EDUC_LEVELS, n, p=[0.30, 0.40, 0.10, 0.20])
    p_lf = np.clip(0.95 - 0.0009 * (age - 40) ** 2 - 0.20 * (sex == 2), 0.05, 0.97)
    in_lf = rng.random(n) < p_lf
    unemp = in_lf & (rng.random(n) < np.where(age < 25, 0.13, 0.04))
    status = np.where(~in_lf, 3, np.where(unemp, 2, 1))
    employed = status == 1
    hours = np.where(employed, np.clip(rng.normal(42, 12, n).round(), 2, 96), np.nan)
    normal_day = np.where(employed, np.clip((hours / rng.choice([5, 6], n)).round(), 1, 16), np.nan)
    wclass = np.where(employed, rng.choice(list(WORKER_CLASS), n, p=[.05, .45, .10, .27, .03, .02, .08]), np.nan)
    wage_worker = np.isin(wclass, [0, 1, 2, 5])
    basis = np.where(wage_worker, rng.choice([3, 4, 1, 5], n, p=[.55, .35, .05, .05]), np.nan)
    mult = np.select([region == 13, region == 4, region == 15], [1.45, 1.15, 0.70], 1.0)
    mult = mult * np.select([educ == EDUC_LEVELS[3], educ == EDUC_LEVELS[0]], [1.5, 0.8], 1.0)
    base = 150 + 18 * age - 0.2 * age ** 2
    pay = np.where(wage_worker, (base * mult * rng.lognormal(0, 0.3, n)).round(), np.nan)
    out = pd.DataFrame({"person_id": np.arange(1, n + 1), "region_code": region,
                        "region": pd.Series(region).map(REGION_NAMES), "sex_code": sex,
                        "sex": pd.Series(sex).map(SEX), "age": age})
    out["marital"] = np.where(age < 25, "Single", rng.choice(["Single", "Married", "Widowed"], n, p=[.3, .62, .08]))
    out["educ"], out["empstat"] = educ, pd.Series(status).map(EMPSTAT)
    out["normal_hrs_day"], out["hours_week"] = normal_day, hours
    out["total_hours_week"] = np.where(employed, hours + rng.choice([0, 0, 0, 8], n), np.nan)
    out["worker_class"] = pd.Series(wclass).map(WORKER_CLASS)
    out["pay_basis"] = pd.Series(basis).map(PAY_BASIS)
    out["daily_pay"], out["weight"] = pay, rng.uniform(150, 900, n).round(2)
    return out

print("✅ Helpers ready: check(), number_line(), venn2_counts(), make_practice_data()")
````

## Lesson 0.2 · Variables and arithmetic ⏱️ 8 min
A **variable** is a labelled box that stores a value. The `=` sign means *"put the value on the right into the box on the left"*. It is **not** the mathematical "equals" of an equation.

| Math | Python | Example | Result |
|---|---|---|---|
| $a + b$ | `a + b` | `7 + 2` | `9` |
| $a - b$ | `a - b` | `7 - 2` | `5` |
| $a \times b$ | `a * b` | `7 * 2` | `14` |
| $a \div b$ | `a / b` | `7 / 2` | `3.5` |
| $a^b$ | `a ** b` | `2 ** 3` | `8` ⚠️ *not* `2 ^ 3` |
| whole-number part of $a \div b$ | `a // b` | `7 // 2` | `3` |
| remainder | `a % b` | `7 % 2` | `1` |

````python
daily_pay = 500          # a worker earns ₱500 per day (a made-up example)
days_worked = 6          # she works 6 days this week
weekly_pay = daily_pay * days_worked
print(weekly_pay)
````

**Order of operations (PEMDAS) works in Python too:** Parentheses → Exponents → Multiplication/Division → Addition/Subtraction.
✍️ *Predict the four outputs on paper first.*

````python
print(2 + 3 * 4)         # multiplication before addition → 14
print((2 + 3) * 4)       # parentheses first → 20
print(-3 ** 2)           # ⚠️ the power is done BEFORE the minus sign → -9
print((-3) ** 2)         # → 9
````

> ⚠️ `-3 ** 2` gives `-9` in Python, just as $-3^2 = -(3^2) = -9$ in algebra. If you mean "negative three, squared", write `(-3) ** 2`.

## 🧠 Going further (optional): what are survey weights?
The LFS does not interview every Filipino. Each sampled person **represents** many others, and that number is their **weight** (`PWGTPRV`, stored here as `weight`). Some groups are sampled more heavily than others. So an *unweighted* average (every row counts once) can differ from a *weighted* average (each row counts `weight` times):
$$\bar{x}_w = \frac{\sum_i w_i x_i}{\sum_i w_i}.$$
Official PSA statistics are **weighted**. In this notebook we keep things **unweighted** for simplicity. That is one reason our numbers are only *teaching estimates*.

````python
unweighted = paid["daily_pay"].mean()
weighted = np.average(paid["daily_pay"], weights=paid["weight"])   # Σ w·x / Σ w
print(f"Unweighted mean daily pay: ₱{unweighted:,.2f}")
print(f"Weighted mean daily pay:   ₱{weighted:,.2f}")
````

## Lesson 1.2 · Average rate of change = slope of a secant line ⏱️ 10 min

### 📖 Step 1 · Read & Understand: average rate of change

> **Concept card**
>
> For a function $f$ between $x = a$ and $x = b$:
> $$\text{average rate of change} = \frac{f(b) - f(a)}{b - a} = \frac{\Delta f}{\Delta x}.$$
> This is the slope of the **secant line** through $(a, f(a))$ and $(b, f(b))$.
> *Intuition:* "on average, how much did the output change per one unit of input?" 
>
> 📚 *Read more:* Sydsaeter et al., *Differentiation*: slopes of curves / rates of change.

### ✍️ Step 2 · Solve by Hand
Suppose average daily pay is ₱400 at age 25 and ₱520 at age 35. Then
$$\frac{520 - 400}{35 - 25} = \frac{120}{10} = 12 \text{ pesos per year of age.}$$
### 🐍 Step 3 · Check with Python

````python
pay_25, pay_35 = 400, 520
print("Average rate of change:", (pay_35 - pay_25) / (35 - 25), "pesos per year")
````

### 📊 Step 4 · Apply to LFS Data
**Question:** how does average daily pay change from one age group to the next?
We group workers aged 20–69 by **decade** using `(age // 10) * 10` (so 27 → 20, 43 → 40). Then `groupby` computes the mean pay in each group. `groupby` is the one new Python idea in this lesson: *"split the table into groups, then summarize each group"*.

````python
adults = paid[paid["age"] >= 20]                  # ages 20 and above ...
adults = adults[adults["age"] < 70]               # ... and below 70
decade = (adults["age"] // 10) * 10               # 20, 30, 40, 50, 60
pay_by_decade = adults.groupby(decade)["daily_pay"].mean()
print(pay_by_decade.round(2))
````

````python
slopes = pay_by_decade.diff() / 10                # (change in mean pay) ÷ (10 years)
print("Average rate of change of pay (₱ per year of age), from the previous decade:")
print(slopes.round(2))
````

````python
fig, ax = plt.subplots()
ax.plot(pay_by_decade.index + 5, pay_by_decade.values, "o-", ms=9, lw=2)   # +5 → middle of each decade
ax.set_title("Mean daily pay by age group, with secant lines (LFS 2018 sample)")
ax.set_xlabel("Age (middle of each 10-year group)"); ax.set_ylabel("Mean daily pay (₱)")
plt.show()
````

*Alt text:* dots for the mean daily pay of the 20s, 30s, 40s, 50s and 60s, joined by straight segments (secant lines).

## Lesson 1.3 · Instantaneous rate of change: the derivative idea ⏱️ 12 min

### 📖 Step 1 · Read & Understand: instantaneous rate of change (derivative)

> **Concept card**
>
> Make the interval $[a, a+h]$ **shorter and shorter** ($h \to 0$). If the average rates of change settle down to one number, that number is the **instantaneous rate of change**, or **derivative**:
> $$f'(a) = \lim_{h \to 0} \frac{f(a+h) - f(a)}{h}.$$
> "$\lim$" (**limit**) means "the value the expression approaches". You will study limits formally in Weeks 11–13.
> *Intuition:* the slope of the curve **at a single point** (the tangent line).
>
> 📚 *Read more:* Sydsaeter et al., *Differentiation*: the derivative / tangent lines; Chiang & Wainwright, *The Concept of Derivative*.

### ✍️ Step 2 · Solve by Hand
Let $f(x) = x^2$ and $a = 3$, so $f(3) = 9$.
- $h = 1$: $\dfrac{f(4) - f(3)}{1} = \dfrac{16 - 9}{1} = 7$
- $h = 0.1$: $\dfrac{f(3.1) - f(3)}{0.1} = \dfrac{9.61 - 9}{0.1} = 6.1$
- $h = 0.01$: $\dfrac{9.0601 - 9}{0.01} = 6.01$

The values approach **6**. (Algebra shows why: $\frac{(3+h)^2 - 9}{h} = \frac{6h + h^2}{h} = 6 + h \to 6$. You'll use that *expand-then-cancel* trick constantly in Parts 5–6!)
### 🐍 Step 3 · Check with Python

````python
def f(x):
    return x ** 2

for h in [1, 0.1, 0.01, 0.001]:
    slope = (f(3 + h) - f(3)) / h
    print(f"h = {h:<6}  average rate of change = {slope:.4f}")
````

### 📊 Step 4 · Apply to LFS Data
Real data is noisy, so we first fit a smooth curve through it. `np.polyfit(x, y, 2)` finds the **best-fitting quadratic** $p(a) = c_0 + c_1 a + c_2 a^2$, the *age–pay profile* (you'll study polynomials in Part 5).

````python
c2, c1, c0 = np.polyfit(paid["age"], paid["daily_pay"], 2)   # coefficients, highest power first
print(f"Fitted profile: p(a) = {c0:.2f} + {c1:.3f}·a + ({c2:.5f})·a²")

def p(age_value):
    """Predicted daily pay (₱) at a given age, from the fitted quadratic."""
    return c0 + c1 * age_value + c2 * age_value ** 2
````

````python
ages_grid = np.linspace(15, paid["age"].max(), 200)
fig, ax = plt.subplots()
ax.scatter(paid["age"], paid["daily_pay"], s=8, alpha=0.25, label="Workers (sample)")
ax.plot(ages_grid, p(ages_grid), lw=3, color=OKABE_ITO[3], label="Fitted quadratic p(a)")
ax.set_title("Daily pay vs. age with a fitted quadratic profile (LFS 2018 sample)")
ax.set_xlabel("Age (years)"); ax.set_ylabel("Daily pay (₱)"); ax.legend(); plt.show()
````

*Alt text:* a cloud of dots (one per worker) with a smooth orange curve through the middle.
**Takeaway:** the curve summarizes the typical pay at each age. Its *steepness* changes with age, so the rate of change is different at different ages.

Now the limit idea on real data: the slope of $p$ at age 30, over shrinking intervals.

````python
a0 = 30
for h in [1, 0.1, 0.01, 0.001, 0.0001]:
    slope = (p(a0 + h) - p(a0)) / h
    print(f"h = {h:<7} slope over [30, 30+h] = {slope:.5f} pesos per year")
````

The numbers **settle down**. That settling value is the derivative $p'(30)$. SymPy can compute derivatives exactly with `sp.diff`. Let's confirm:

````python
a = sp.symbols("a")                      # the letter a (age) as a SymPy symbol
p_sym = c0 + c1 * a + c2 * a ** 2        # the same profile, written with the symbol
dp = sp.diff(p_sym, a)                   # derivative with respect to a
print("p'(a) =", dp)
print("p'(30) =", round(float(dp.subs(a, 30)), 5), "pesos per year")
````

## Lesson 1.4 · Marginal thinking ⏱️ 5 min
In economics, a derivative is often called a **marginal** quantity: *marginal cost, marginal revenue, marginal product*. Here, $p'(a)$ is the **marginal pay of one more year of age** (a rough stand-in for experience).
> 🧠 Marginal thinking is one of the key ideas in the economics program ("opportunity cost, marginal, and positive-sum thinking"). Calculus is its language.

````python
for age_value in [20, 30, 40, 50, 60]:
    print(f"Age {age_value}: marginal pay p'(a) = {float(dp.subs(a, age_value)):8.2f} ₱ per year")
````

## Lesson 2.1 · Sets and set operations ⏱️ 20 min

### 📖 Step 1 · Read & Understand: sets

> **Concept card**
>
> A **set** is a well-defined collection of objects called **elements**. We write $x \in A$ ("$x$ is in $A$").
> - **Universal set** $U$: everything under discussion. **Subset** $A \subseteq B$: every element of $A$ is in $B$.
> - **Union** $A \cup B$: in $A$ **or** $B$ (or both). **Intersection** $A \cap B$: in **both**.
> - **Complement** $A'$ (or $A^c$): in $U$ but **not** in $A$. **Difference** $A - B$: in $A$ but not in $B$.
> - **Cardinality** $|A|$: the number of elements.
> *Intuition:* sets are groups; the operations are "or", "and", "not", and "but not".
>
> 📚 *Read more:* Swokowski & Cole, *Real Numbers* (sets and notation); Sydsaeter et al., *Essentials of Logic and Set Theory*.

### ✍️ Step 2 · Solve by Hand
Eight surveyed persons are numbered 1–8, so $U = \{1, 2, \dots, 8\}$. Let $A = \{1, 2, 3, 5, 8\}$ be the employed and $B = \{2, 3, 4, 6, 8\}$ the females.
- $A \cap B = \{2, 3, 8\}$ (elements in **both**): *employed women*.
- $A \cup B = \{1, 2, 3, 4, 5, 6, 8\}$ (in **at least one**).
- $A' = U - A = \{4, 6, 7\}$: *not employed*.
- $A - B = \{1, 5\}$: *employed and not female*.
- $\{2, 3\} \subseteq A$? Yes, because both 2 and 3 are in $A$.

### 🐍 Step 3 · Check with Python
Python has a built-in **set** type, written with curly braces. This is the new Python idea of this lesson. The operators are `&` (∩), `|` (∪), `-` (difference), and `<=` (⊆).

````python
U = {1, 2, 3, 4, 5, 6, 7, 8}
A = {1, 2, 3, 5, 8}          # employed
B = {2, 3, 4, 6, 8}          # female
print("A ∩ B =", A & B)
print("A ∪ B =", A | B)
print("A'    =", U - A)
print("A − B =", A - B)
print("{2, 3} ⊆ A ?", {2, 3} <= A)
````

✅ The output matches the hand solution.

## Lesson 2.2 · De Morgan's laws and counting with inclusion–exclusion ⏱️ 15 min

### 📖 Step 1 · Read & Understand: De Morgan's laws and inclusion–exclusion

> **Concept card**
>
> **De Morgan's laws:** $(A \cup B)' = A' \cap B'$ and $(A \cap B)' = A' \cup B'$.
> "Not (A or B)" = "not A **and** not B"; "not (A and B)" = "not A **or** not B".
> **Inclusion–exclusion:** $|A \cup B| = |A| + |B| - |A \cap B|$. Adding $|A|$ and $|B|$ counts the overlap twice, so subtract it once.
>
> 📚 *Read more:* Sydsaeter et al., *Essentials of Logic and Set Theory* (set operations, Venn diagrams); Swokowski & Cole, *Real Numbers*.

### ✍️ Step 2 · Solve by Hand (same $U, A, B$ as Lesson 2.1)
- $(A \cup B)' = U - \{1,2,3,4,5,6,8\} = \{7\}$, and $A' \cap B' = \{4,6,7\} \cap \{1,5,7\} = \{7\}$. ✔️ Equal, as De Morgan says.
- $|A \cup B| = |A| + |B| - |A \cap B| = 5 + 5 - 3 = 7$. ✔️ Matches $|\{1,2,3,4,5,6,8\}| = 7$.
### 🐍 Step 3 · Check with Python

````python
print("(A ∪ B)' =", U - (A | B), "  A' ∩ B' =", (U - A) & (U - B))
print("De Morgan holds?", (U - (A | B)) == ((U - A) & (U - B)))
print("|A ∪ B| =", len(A | B), "  |A| + |B| − |A ∩ B| =", len(A) + len(B) - len(A & B))
````

### 📊 Step 4 · Apply to LFS Data
**New Python idea: boolean masks.** Instead of building sets of IDs, pandas can combine True/False columns directly:
`&` = and (∩), `|` = or (∪), `~` = not (complement). **Always wrap each condition in parentheses.** `.sum()` counts the `True` values, because True counts as 1 and False as 0.

````python
is_emp = df["empstat"] == "Employed"      # True for employed persons
is_fem = df["sex"] == "Female"            # True for women
is_ncr = df["region"] == "NCR"            # True for NCR residents
print("Employed AND female :", (is_emp & is_fem).sum())
print("Employed OR NCR     :", (is_emp | is_ncr).sum())
print("NOT employed        :", (~is_emp).sum())
````

````python
# De Morgan on real data: (E ∪ N)' versus E' ∩ N'
left = ~(is_emp | is_ncr)
right = (~is_emp) & (~is_ncr)
print("Count of (E ∪ N)':", left.sum(), "| count of E' ∩ N':", right.sum())
print("Same people, row by row?", (left == right).all())
# Inclusion–exclusion on real data
print("|E ∪ N| =", (is_emp | is_ncr).sum(), "= |E| + |N| − |E ∩ N| =", is_emp.sum() + is_ncr.sum() - (is_emp & is_ncr).sum())
````

## Lesson 2.3 · Types of numbers (and why codes are not quantities) ⏱️ 15 min

### 📖 Step 1 · Read & Understand: the real number system

> **Concept card**
>
> - **Natural numbers** $\mathbb{N} = \{1, 2, 3, \dots\}$ (some books include 0; check your textbook's convention).
> - **Integers** $\mathbb{Z} = \{\dots, -2, -1, 0, 1, 2, \dots\}$.
> - **Rational numbers** $\mathbb{Q}$: numbers $p/q$ with $p, q \in \mathbb{Z}$, $q \neq 0$. Their decimals **terminate or repeat** (0.75, 0.333…).
> - **Irrational numbers**: real numbers that are **not** rational, with non-repeating, non-terminating decimals ($\sqrt{2}$, $\pi$).
> - **Real numbers** $\mathbb{R}$ = rationals ∪ irrationals: every point on the number line.
> $$\mathbb{N} \subset \mathbb{Z} \subset \mathbb{Q} \subset \mathbb{R}.$$
> *Intuition:* each set "fills in gaps" left by the one before it.
>
> 📚 *Read more:* Swokowski & Cole, *Real Numbers*; Rees, Sparks & Rees, *The Number System*; Chiang & Wainwright, *Economic Models* (the real-number system).

### ✍️ Step 2 · Solve by Hand
Classify each number (list the **smallest** set it belongs to):

| Number | Smallest set | Why |
|---|---|---|
| $\sqrt{9}$ | $\mathbb{N}$ | $\sqrt{9} = 3$ |
| $0$ | $\mathbb{Z}$ | an integer (it is in $\mathbb{N}$ only if your book includes 0) |
| $-4$ | $\mathbb{Z}$ | negative whole number |
| $3/4 = 0.75$ | $\mathbb{Q}$ | ratio of integers; terminating decimal |
| $0.333\ldots$ | $\mathbb{Q}$ | repeating decimal $= 1/3$ |
| $537.50$ | $\mathbb{Q}$ | $= 1075/2$ |
| $\sqrt{2}$, $\pi$ | irrational | never terminates or repeats |

### 🐍 Step 3 · Check with Python
SymPy works with **exact** numbers. `sp.Rational(1, 3)` is exactly $1/3$. `sp.sqrt(9)` simplifies to 3, but `sp.sqrt(2)` stays as $\sqrt{2}$ because it can't be written as a fraction.

````python
print(sp.sqrt(9), "| rational?", sp.sqrt(9).is_rational)
print(sp.sqrt(2), "| rational?", sp.sqrt(2).is_rational)
print(sp.pi, "| rational?", sp.pi.is_rational)
print(sp.nsimplify(0.75), "and", sp.nsimplify(537.50))   # turn a decimal into a fraction
````

**⚠️ Computers store most decimals approximately (the new Python idea of this lesson).** Python's `float` can't store $0.1$ exactly, so tiny errors appear. Compare floats with `round` or `np.isclose`, never with `==`.

````python
print(0.1 + 0.2)                       # not exactly 0.3!
print(0.1 + 0.2 == 0.3)                # False, because of tiny storage errors
print(round(0.1 + 0.2, 10) == 0.3)     # True after rounding
print(np.isclose(0.1 + 0.2, 0.3))      # True: "equal up to a tiny tolerance"
````

## Lesson 2.4 · Properties of real numbers ⏱️ 15 min

### 📖 Step 1 · Read & Understand: field properties of ℝ

> **Concept card**
>
> For all real $a, b, c$:
> | Property | Addition | Multiplication |
> |---|---|---|
> | Commutative | $a + b = b + a$ | $ab = ba$ |
> | Associative | $(a + b) + c = a + (b + c)$ | $(ab)c = a(bc)$ |
> | Identity | $a + 0 = a$ | $a \cdot 1 = a$ |
> | Inverse | $a + (-a) = 0$ | $a \cdot \frac{1}{a} = 1$ **if $a \neq 0$** |
> | Distributive | $a(b + c) = ab + ac$ | |
>
> *Intuition:* these are the "legal moves" of algebra. Every simplification step is one of them.
>
> 📚 *Read more:* Swokowski & Cole, *Real Numbers* (properties of real numbers); Sydsaeter et al., *Algebra* (rules of algebra).

### ✍️ Step 2 · Solve by Hand (board-work style: name every property)
A worker gets ₱450 basic pay plus a ₱50 allowance per day for 5 days. Total weekly pay:
$$
\begin{aligned}
5(450 + 50) &= 5 \cdot 450 + 5 \cdot 50 && \text{distributive property} \\
&= 2250 + 250 && \text{multiplication} \\
&= 2500. &&
\end{aligned}
$$
Simplify $4(x + 5) + 3x$:
$$
\begin{aligned}
4(x + 5) + 3x &= (4x + 20) + 3x && \text{distributive} \\
&= 4x + (20 + 3x) && \text{associative (+)} \\
&= 4x + (3x + 20) && \text{commutative (+)} \\
&= (4x + 3x) + 20 && \text{associative (+)} \\
&= (4 + 3)x + 20 = 7x + 20 && \text{distributive (factoring out } x).
\end{aligned}
$$
### 🐍 Step 3 · Check with Python

````python
print(5 * (450 + 50), "=", 5 * 450 + 5 * 50)          # distributive, with numbers
x = sp.symbols("x")
print(sp.expand(4 * (x + 5) + 3 * x))                 # SymPy simplifies to 7x + 20
````

### 📊 Step 4 · Apply to LFS Data
**Question:** if every paid worker in our sample worked 5 days, what would the *total* weekly pay bill be? Distributivity says that **multiplying each pay by 5 and then adding** gives the same result as **adding first and then multiplying by 5**: $\sum 5d_i = 5\sum d_i$.

````python
way_1 = (5 * paid["daily_pay"]).sum()                  # multiply each pay by 5, then add
way_2 = 5 * paid["daily_pay"].sum()                    # add first, then multiply by 5
print(f"Way 1: ₱{way_1:,.2f}   Way 2: ₱{way_2:,.2f}")
print("Equal (up to float rounding)?", np.isclose(way_1, way_2))
````

````python
# Commutativity: adding in a different order (sorted from smallest to largest) gives the same total
print("Original order:", paid["daily_pay"].sum(), "| sorted order:", paid["daily_pay"].sort_values().sum())
# Multiplicative inverse: d × (1/d) = 1 for every pay, which works only because no pay is 0 (we set 0 → NaN)
print("d·(1/d) = 1 for everyone?", np.allclose(paid["daily_pay"] * (1 / paid["daily_pay"]), 1))
````

## ⚠️ Common mistakes
- Writing `df["empstat"] == "Employed" & df["sex"] == "Female"` **without parentheses**. Python then evaluates `&` first, and you get an error or wrong results. ✅ `(cond1) & (cond2)`.
- Thinking $(A \cap B)' = A' \cap B'$. ❌ By De Morgan, $(A \cap B)' = A' \cup B'$.
- Forgetting to subtract the overlap: $|A \cup B| \neq |A| + |B|$ unless $A \cap B = \varnothing$.
- Averaging codes (`region_code`, `sex_code`). They are **labels**.
- Assuming subtraction or division is commutative: $5 - 3 \neq 3 - 5$, $6/3 \neq 3/6$.
- Testing floats with `==`: `0.1 + 0.2 == 0.3` is `False`. Use `np.isclose`.
- Saying "$\frac{1}{a}$ is the inverse of $a$" without the condition **$a \neq 0$**.

## Lesson 3.1 · Order properties and linear inequalities ⏱️ 20 min

### 📖 Step 1 · Read & Understand: order properties

> **Concept card**
>
> For real $a, b, c$ with $a < b$:
> 1. $a + c < b + c$ (adding or subtracting keeps the direction).
> 2. If $c > 0$: $ac < bc$ (multiplying by a **positive** number keeps the direction).
> 3. If $c < 0$: $ac > bc$ (multiplying or dividing by a **negative** number **reverses** the direction).
> A **linear inequality** like $mx + k \ge q$ is solved like an equation, while watching rule 3.
> *Intuition:* multiplying by $-1$ mirrors the number line, so "left of" becomes "right of".
>
> 📚 *Read more:* Swokowski & Cole, *Inequalities*; Sydsaeter et al., *Algebra* (inequalities).

### ✍️ Step 2 · Solve by Hand
**(a)** A worker paid ₱62.50 per hour wants weekly earnings of **at least** ₱2,500:
$$62.5h \ge 2500 \;\Longrightarrow\; h \ge \frac{2500}{62.5} = 40 \quad \text{(divided by } 62.5 > 0\text{: direction kept)}.$$
**(b)** A ₱500 weekly transport allowance shrinks by ₱20 for every hour of overtime $h$. When does more than ₱100 remain?
$$500 - 20h > 100 \;\Rightarrow\; -20h > -400 \;\Rightarrow\; h < 20 \quad \text{(divided by } -20 < 0\text{: direction **reversed**)}.$$
### 🐍 Step 3 · Check with Python
The new Python idea: `sp.solveset(inequality, variable, domain=sp.S.Reals)` returns the solution **as an interval**. `sp.Rational("62.5")` keeps 62.5 exact ($=125/2$).

````python
h = sp.symbols("h", real=True)
print(sp.solveset(sp.Rational("62.5") * h >= 2500, h, domain=sp.S.Reals))   # expect [40, ∞)
print(sp.solveset(500 - 20 * h > 100, h, domain=sp.S.Reals))                # expect (−∞, 20)
````

`Interval(40, oo)` means $[40, \infty)$ and `Interval.open(-oo, 20)` means $(-\infty, 20)$. ✅ Matches.

## Lesson 3.2 · Compound inequalities and interval notation ⏱️ 20 min

### 📖 Step 1 · Read & Understand: intervals

> **Concept card**
>
> | Inequality | Interval | Type |
> |---|---|---|
> | $a < x < b$ | $(a, b)$ | open |
> | $a \le x \le b$ | $[a, b]$ | closed |
> | $a \le x < b$ | $[a, b)$ | half-open |
> | $x \ge a$ | $[a, \infty)$ | unbounded |
> | $x < b$ | $(-\infty, b)$ | unbounded |
> A square bracket **includes** the endpoint (●); a round bracket **excludes** it (○). $\infty$ is not a number, so it always gets a round bracket.
> A **compound inequality** $p \le mx + k < q$ is solved by doing the same operation to **all three parts**.
>
> 📚 *Read more:* Swokowski & Cole, *Real Numbers* (intervals) and *Inequalities* (compound inequalities); Sydsaeter et al., *Algebra* (intervals).

### ✍️ Step 2 · Solve by Hand
**(a)** $-3 \le 2x - 5 < 7 \;\Rightarrow\; 2 \le 2x < 12$ (add 5 to all parts) $\;\Rightarrow\; 1 \le x < 6$ (divide all parts by $2 > 0$). Solution: $[1, 6)$.
**(b)** The working-age population is $15 \le \text{age} \le 64$, or $[15, 64]$. Persons aged 25 and over: $[25, \infty)$. Their **intersection** is $[15, 64] \cap [25, \infty) = [25, 64]$.
**(c)** "Dependent ages": youth or elderly, $[15, 24] \cup [65, \infty)$. This union can't be written as a single interval.

### 🐍 Step 3 · Check with Python

````python
x = sp.symbols("x", real=True)
print(sp.reduce_inequalities([2 * x - 5 >= -3, 2 * x - 5 < 7], x))   # both parts at once
working_age = sp.Interval(15, 64)                                     # closed [15, 64]
print(working_age.intersect(sp.Interval(25, sp.oo)))                  # [25, 64]
print(sp.Interval(15, 24).union(sp.Interval(65, sp.oo)))
print("Is 64.5 in [15, 64]?", working_age.contains(64.5))
````

## ⚠️ Common mistakes
- **Not flipping** the inequality when multiplying or dividing by a negative: $-2x > 6 \Rightarrow x < -3$ (not $x > -3$).
- Writing $[5, \infty]$. ❌ Infinity is never included: $[5, \infty)$.
- Mixing up brackets: $(0, 40)$ excludes 0 and 40; $[0, 40]$ includes both.
- In a compound inequality, applying an operation to only **two** of the three parts.
- Thinking `.between(a, b)` is open. By default it is **closed**, $[a, b]$.
- Forgetting `right=False` in `pd.cut`. The default makes bins $(a, b]$, so a 25-year-old would land in the wrong group.

## Lesson 4.1 · Laws of exponents and compound growth ⏱️ 20 min

### 📖 Step 1 · Read & Understand: laws of exponents

> **Concept card**
>
> For real $a, b$ and integers $m, n$ (with $a, b \neq 0$ wherever we divide or use exponents $\le 0$):
> | Law | Statement |
> |---|---|
> | Product | $a^m a^n = a^{m+n}$ |
> | Quotient | $a^m / a^n = a^{m-n}$ |
> | Power of a power | $(a^m)^n = a^{mn}$ |
> | Power of a product / quotient | $(ab)^n = a^n b^n$, $(a/b)^n = a^n / b^n$ |
> | Zero exponent | $a^0 = 1$ ($a \neq 0$) |
> | Negative exponent | $a^{-n} = 1/a^n$ ($a \neq 0$) |
> **Compound growth:** a value $P$ growing at rate $r$ per period for $t$ periods becomes $P(1 + r)^t$.
>
> 📚 *Read more:* Swokowski & Cole, *Exponents and Radicals*; Sydsaeter et al., *Algebra* (powers; compound interest).

### ✍️ Step 2 · Solve by Hand
- $\dfrac{2^3 \cdot 2^4}{2^5} = \dfrac{2^{7}}{2^5}$ (product law) $= 2^{2}$ (quotient law) $= 4$.
- $(3x^2)^3 = 3^3 (x^2)^3$ (power of a product) $= 27x^6$ (power of a power).
- $5^{-2} = \dfrac{1}{5^2} = \dfrac{1}{25}$ (negative exponent).
- **Growth:** ₱500 growing 5% per year for 2 years gives $500(1.05)^2 = 500(1.1025) = 551.25$.
### 🐍 Step 3 · Check with Python

````python
print((2 ** 3 * 2 ** 4) / 2 ** 5)          # 4.0
print(5 ** -2, "=", 1 / 25)                 # negative exponent
x = sp.symbols("x", positive=True)
print(sp.expand((3 * x ** 2) ** 3))          # 27*x**6
print(500 * 1.05 ** 2)                       # 551.25
````

**Question (hypothetical):** if the median daily pay in our sample grew 5% per year, what would it be after $t = 0, 1, \dots, 10$ years?
New Python idea: `np.arange(0, 11)` makes the array `0, 1, …, 10`, and `**` then works on **every element at once**.

````python
P = paid["daily_pay"].median()               # starting value: median daily pay in our sample
r = 0.05                                     # hypothetical 5% growth per year
t = np.arange(0, 11)                         # 0, 1, 2, ..., 10 years
projected = P * (1 + r) ** t                 # compound growth, element by element
for year, value in zip(t, projected):        # zip pairs up the two lists
    print(f"Year {year:2d}: ₱{value:,.2f}")
````

## Lesson 4.2 · Radicals and rational exponents ⏱️ 20 min

### 📖 Step 1 · Read & Understand: radicals and rational exponents

> **Concept card**
>
> $\sqrt[n]{a} = a^{1/n}$ and $a^{m/n} = \left(\sqrt[n]{a}\right)^m = \sqrt[n]{a^m}$.
> **Domain:** for **even** $n$, $\sqrt[n]{a}$ is real only when $a \ge 0$. For odd $n$, any real $a$ is fine.
> $\sqrt{ab} = \sqrt{a}\sqrt{b}$ for $a, b \ge 0$. **But** $\sqrt{a + b} \neq \sqrt{a} + \sqrt{b}$ in general!
> **Rates:** an annual rate $r$ is equivalent to the monthly rate $(1 + r)^{1/12} - 1$. The constant yearly rate that takes $P$ to $Q$ in $t$ years is $r = (Q/P)^{1/t} - 1$.
>
> 📚 *Read more:* Swokowski & Cole, *Exponents and Radicals* (rational exponents, simplifying radicals); Rees, Sparks & Rees, *Exponents and Radicals*; Sydsaeter et al., *Algebra*.

### ✍️ Step 2 · Solve by Hand
- $8^{2/3} = (\sqrt[3]{8})^2 = 2^2 = 4$.
- $16^{-3/4} = \dfrac{1}{(\sqrt[4]{16})^3} = \dfrac{1}{2^3} = \dfrac{1}{8}$.
- $\sqrt{72} = \sqrt{36 \cdot 2} = 6\sqrt{2}$.
- **Doubling in 10 years:** $(1 + r)^{10} = 2 \Rightarrow 1 + r = 2^{1/10} \Rightarrow r = 2^{1/10} - 1$ (about 7.2%).
### 🐍 Step 3 · Check with Python
Floats give *approximate* answers. SymPy gives **exact** ones:

````python
print(8 ** (2 / 3))                                      # 3.9999999999999996, a float approximation!
print(sp.Integer(8) ** sp.Rational(2, 3))                # exactly 4
print(sp.Integer(16) ** sp.Rational(-3, 4), "|", sp.sqrt(72))   # 1/8 | 6*sqrt(2)
print("Doubling rate:", 2 ** (1 / 10) - 1)
````

> ⚠️ **Python and negative bases.** `(-8) ** (1/3)` does **not** give $-2$. Python returns a complex number, because `1/3` is a float. In this course, take odd roots of negatives by hand: $\sqrt[3]{-8} = -2$. Use `np.cbrt(-8)` or `sp.real_root(-8, 3)` to check.

## Lesson 4.3 · Absolute value and distance ⏱️ 20 min

### 📖 Step 1 · Read & Understand: absolute value and distance

> **Concept card**
>
> $$|a| = \begin{cases} a & \text{if } a \ge 0 \\ -a & \text{if } a < 0 \end{cases}$$
> $|a|$ is the distance from $a$ to 0, so it is never negative. The **distance** between $a$ and $b$ on the number line is $d(a, b) = |a - b| = |b - a|$.
> Properties: $|ab| = |a||b|$, $|a/b| = |a|/|b|$ ($b \neq 0$), and the triangle inequality $|a + b| \le |a| + |b|$.
> **Mean absolute deviation** from the median $m$: $\frac{1}{n}\sum |x_i - m|$, the average distance of the data from its center.
>
> 📚 *Read more:* Swokowski & Cole, *Real Numbers* (absolute value; distance between points on a coordinate line); Sydsaeter et al., *Algebra* (absolute value).

### ✍️ Step 2 · Solve by Hand
- $|-7| = 7$; $|3 - 10| = |-7| = 7$.
- $d(-3, 5) = |-3 - 5| = |-8| = 8$, and also $|5 - (-3)| = 8$ (symmetric).
- Two regions' median pays of ₱450 and ₱610 are $|450 - 610| = 160$ pesos apart.
- The data $\{400, 500, 650\}$ have median $500$. MAD $= \frac{|400-500| + |500-500| + |650-500|}{3} = \frac{100 + 0 + 150}{3} = 83.\overline{3}$.
### 🐍 Step 3 · Check with Python
New Python idea: `abs()` for one number, `np.abs()` for a whole array or column.

````python
print(abs(-7), abs(3 - 10), abs(-3 - 5), abs(450 - 610))
sample = np.array([400, 500, 650])
print("MAD:", np.abs(sample - np.median(sample)).mean())
````

## ⚠️ Common mistakes
- $\sqrt{a + b} \neq \sqrt{a} + \sqrt{b}$: e.g. $\sqrt{9 + 16} = 5$, but $3 + 4 = 7$.
- $(a + b)^2 \neq a^2 + b^2$, and $(a+b)^n \ne a^n + b^n$ in general.
- $a^m \cdot a^n \neq a^{mn}$. **Add** the exponents: $a^{m+n}$.
- $-3^2 = -9$ but $(-3)^2 = 9$. Python agrees: `-3 ** 2 == -9`.
- $a^{-n}$ is **not** negative. It is $1/a^n$.
- Taking $\sqrt{x}$ of a negative in the reals, which is undefined. Always state $x \ge 0$.
- $|a - b| \neq |a| - |b|$: e.g. $|3 - (-5)| = 8$, but $|3| - |-5| = -2$.
- Converting an annual rate to monthly as $r/12$ when the question is about compounding.

## Lesson 5.2 · Adding, subtracting and multiplying polynomials ⏱️ 20 min

### 📖 Step 1 · Read & Understand: polynomial operations

> **Concept card**
>
> - **Add/subtract**: combine **like terms** (same power): $(a x^2 + bx) + (c x^2 + dx) = (a + c)x^2 + (b + d)x$.
> - **Multiply**: use the **distributive property** so every term multiplies every term, then combine like terms. For binomials this is FOIL (First, Outer, Inner, Last).
> - The degree of a product is the **sum** of the degrees.
>
> 📚 *Read more:* Swokowski & Cole, *Algebraic Expressions* (operations on polynomials); Rees, Sparks & Rees, *Fundamental Operations*.

### ✍️ Step 2 · Solve by Hand
Let $P = 3x^2 + 2x - 1$ and $Q = x^2 - 5x + 4$.
- $P + Q = (3+1)x^2 + (2-5)x + (-1+4) = 4x^2 - 3x + 3$.
- $P - Q = (3-1)x^2 + (2+5)x + (-1-4) = 2x^2 + 7x - 5$. ⚠️ Distribute the minus sign to **every** term of $Q$.
- $(x + 2)(x - 3) = x^2 - 3x + 2x - 6 = x^2 - x - 6$ (distributive, then like terms).
### 🐍 Step 3 · Check with Python

### 📊 Step 4 · Apply to LFS Data
We fit **separate** quadratic age–pay profiles for men and women, $p_M(a)$ and $p_F(a)$. Then:
- **Subtracting** polynomials gives the predicted **gender pay gap** at each age: $g(a) = p_M(a) - p_F(a)$.
- **Multiplying by a constant** gives weekly profiles: $5\,p(a)$.

New Python idea: `sp.lambdify(a, expr)` turns a SymPy expression into a normal Python function that works on whole arrays, which is handy for plotting.

````python
a = sp.symbols("a")                                   # age

def fit_profile(data):
    """Fit pay = c0 + c1·a + c2·a² to a table with 'age' and 'daily_pay'; return a SymPy expression."""
    c2, c1, c0 = np.polyfit(data["age"], data["daily_pay"], 2)
    return c0 + c1 * a + c2 * a ** 2

p_all = fit_profile(paid)
p_M = fit_profile(paid[paid["sex"] == "Male"])
p_F = fit_profile(paid[paid["sex"] == "Female"])
print("p_M(a) =", sp.N(p_M, 5)); print("p_F(a) =", sp.N(p_F, 5))
````

````python
gap = sp.expand(p_M - p_F)                            # subtracting polynomials: like terms combine
weekly_profile = sp.expand(5 * p_all)                 # scalar multiple
print("Gap g(a) = p_M − p_F =", sp.N(gap, 5))
print("Weekly profile 5·p(a) =", sp.N(weekly_profile, 5))
for age_value in [25, 40, 55]:
    print(f"Age {age_value}: predicted gap ₱{float(gap.subs(a, age_value)):,.2f} per day")
````

## Lesson 5.3 · Special products ⏱️ 20 min

### 📖 Step 1 · Read & Understand: special products

> **Concept card**
>
> | Name | Identity |
> |---|---|
> | Square of a sum | $(u + v)^2 = u^2 + 2uv + v^2$ |
> | Square of a difference | $(u - v)^2 = u^2 - 2uv + v^2$ |
> | Sum × difference | $(u + v)(u - v) = u^2 - v^2$ |
> | Cube of a sum / difference | $(u \pm v)^3 = u^3 \pm 3u^2v + 3uv^2 \pm v^3$ |
> *Intuition:* $(u+v)^2$ is the area of a square of side $u + v$: two squares **plus two rectangles** $uv$.
> **Economics:** two years of growth at rate $g$ multiply by $(1 + g)^2 = 1 + 2g + g^2$. The $g^2$ term is "growth on growth".
>
> 📚 *Read more:* Swokowski & Cole, *Algebraic Expressions* (product formulas); Rees, Sparks & Rees, *Special Products and Factoring*.

### ✍️ Step 2 · Solve by Hand
- $(x + 5)^2 = x^2 + 10x + 25$
- $(2x - 3)^2 = 4x^2 - 12x + 9$
- $(x + 4)(x - 4) = x^2 - 16$
- $(x + 1)^3 = x^3 + 3x^2 + 3x + 1$
- Mental math: $52 \times 48 = (50 + 2)(50 - 2) = 2500 - 4 = 2496$.
- Two years at $g = 5\%$: $(1.05)^2 = 1 + 2(0.05) + 0.05^2 = 1.1025$, so pay rises 10.25%, **not** 10%.

## ⚠️ Common mistakes
- $(a + b)^2 \neq a^2 + b^2$ (missing $2ab$). Likewise $(a - b)^2 \neq a^2 - b^2$.
- Subtracting a polynomial without distributing the minus sign: $x^2 - (3x - 2) = x^2 - 3x + 2$, not $x^2 - 3x - 2$.
- Combining **unlike** terms: $3x^2 + 2x \neq 5x^3$.
- Multiplying exponents when multiplying terms: $x^2 \cdot x^3 = x^5$, not $x^6$.
- Calling $\frac{1}{x} + 2$ a polynomial. Negative exponents are not allowed.
- In Python, writing `2x` instead of `2 * x`, or `x^2` instead of `x ** 2`.

## Lesson 6.1 · Factoring polynomials ⏱️ 25 min

### 📖 Step 1 · Read & Understand: factoring

> **Concept card**
>
> **Factoring** rewrites a sum as a **product**. It is expanding in reverse. Try the methods in this order:
> 1. **Greatest common factor (GCF):** $6x^2 + 9x = 3x(2x + 3)$.
> 2. **Special patterns:** difference of squares $u^2 - v^2 = (u - v)(u + v)$; perfect squares $u^2 \pm 2uv + v^2 = (u \pm v)^2$; sum/difference of cubes $u^3 \pm v^3 = (u \pm v)(u^2 \mp uv + v^2)$.
> 3. **Trinomials** $x^2 + bx + c$: find two numbers whose product is $c$ and whose sum is $b$.
> 4. **Grouping** (four terms): factor pairs, then factor out the common binomial.
> *Always check by expanding.* Note: $u^2 + v^2$ does **not** factor over the reals.
>
> 📚 *Read more:* Swokowski & Cole, *Algebraic Expressions* (factoring formulas); Rees, Sparks & Rees, *Special Products and Factoring*; Dowling, *Review*.

### ✍️ Step 2 · Solve by Hand
- GCF: $6x^2 + 9x = 3x(2x + 3)$.
- Trinomial: $x^2 - 5x + 6$: we need numbers with product $6$ and sum $-5$, which are $-2$ and $-3$. So $(x - 2)(x - 3)$.
- Difference of squares: $x^2 - 49 = (x - 7)(x + 7)$.
- Difference of cubes: $x^3 - 8 = (x - 2)(x^2 + 2x + 4)$.
- Grouping: $xy + 2x + 3y + 6 = x(y + 2) + 3(y + 2) = (x + 3)(y + 2)$.
### 🐍 Step 3 · Check with Python
New Python idea: `sp.factor(...)`, the reverse of `sp.expand`.

````python
x, y = sp.symbols("x y")
for expr in [6 * x**2 + 9 * x, x**2 - 5 * x + 6, x**2 - 49, x**3 - 8, x * y + 2 * x + 3 * y + 6]:
    print(f"{str(expr):<22} =  {sp.factor(expr)}")
print("x² + 4 over the reals:", sp.factor(x**2 + 4), "(doesn't factor)")
````

## Lesson 6.2 · Rational expressions and domain restrictions ⏱️ 25 min

### 📖 Step 1 · Read & Understand: rational expressions

> **Concept card**
>
> A **rational (fractional) expression** is a quotient $\dfrac{P}{Q}$ of polynomials. It is defined only where $Q \neq 0$: that set of $x$ values is its **domain**.
> **Simplify** by factoring top and bottom, then cancelling common **factors** (never terms!):
> $$\frac{ac}{bc} = \frac{a}{b} \quad (b \ne 0,\; c \ne 0).$$
> ⚠️ The restrictions of the **original** expression still apply after cancelling.
>
> 📚 *Read more:* Swokowski & Cole, *Fractional Expressions*; Rees, Sparks & Rees, *Fractions*; Sydsaeter et al., *Algebra* (fractions).

### ✍️ Step 2 · Solve by Hand
$$\frac{x^2 - 9}{x^2 + x - 6} = \frac{(x - 3)(x + 3)}{(x + 3)(x - 2)} = \frac{x - 3}{x - 2}, \qquad x \neq -3,\; x \neq 2.$$
*Why keep $x \ne -3$?* The original expression is undefined at $x = -3$ (it gives $0/0$), even though the simplified one isn't.
### 🐍 Step 3 · Check with Python

````python
expr = (x**2 - 9) / (x**2 + x - 6)
print("Simplified:", sp.cancel(expr))
print("Excluded x (denominator = 0):", sp.solve(sp.denom(expr), x))   # SymPy forgets these when it cancels!
print("Original at x = -3:", expr.subs(x, -3))                        # nan: 0/0 is undefined
````

### 📊 Step 4 · Apply to LFS Data: division by zero in real data
**Question:** what is each worker's **hourly pay**, $\dfrac{\text{daily pay}}{\text{normal hours per day}}$? The domain restriction "denominator $\neq 0$" becomes a data rule: **hours of 0 or missing → no hourly pay**. pandas doesn't stop you from dividing by zero. It quietly returns `inf` (∞):

````python
demo_pay = pd.Series([500.0, 400.0, 600.0])
demo_hours = pd.Series([8.0, 0.0, np.nan])
print(demo_pay / demo_hours)                                   # 62.5, inf, NaN: inf is dangerous!
safe_hours = demo_hours.where(demo_hours > 0)                   # keep hours > 0; anything else → NaN
print(demo_pay / safe_hours)                                   # 62.5, NaN, NaN: safe
````

>
> For nonzero denominators:
> $$\frac{a}{b} \pm \frac{c}{d} = \frac{ad \pm bc}{bd}, \qquad \frac{a}{b}\cdot\frac{c}{d} = \frac{ac}{bd}, \qquad \frac{a}{b} \div \frac{c}{d} = \frac{a}{b}\cdot\frac{d}{c} \;\;(c \neq 0).$$
> Use the **least common denominator (LCD)** to keep things small.
> **Labor-market rates** (with $E$ = employed, $U$ = unemployed, $P$ = population 15+, and labor force $LF = E + U$):
> $$\text{LFPR} = \frac{E + U}{P}, \qquad \text{UR} = \frac{U}{E + U}, \qquad \text{EPR} = \frac{E}{P}.$$
>
> 📚 *Read more:* Swokowski & Cole, *Fractional Expressions* (operations; complex fractions); Rees, Sparks & Rees, *Fractions*.

### ✍️ Step 2 · Solve by Hand
**(a)** $\dfrac{1}{x} + \dfrac{1}{x + 1} = \dfrac{(x + 1) + x}{x(x + 1)} = \dfrac{2x + 1}{x(x + 1)}$, with $x \neq 0, -1$.

**(b) An identity linking the three rates.** Divide EPR by LFPR, which is a complex fraction:
$$\frac{\text{EPR}}{\text{LFPR}} = \frac{E/P}{(E + U)/P} = \frac{E}{P}\cdot\frac{P}{E + U} = \frac{E}{E + U}.$$
So
$$1 - \frac{\text{EPR}}{\text{LFPR}} = \frac{E + U}{E + U} - \frac{E}{E + U} = \frac{U}{E + U} = \text{UR}.$$
### 🐍 Step 3 · Check with Python

````python
print(sp.together(1 / x + 1 / (x + 1)))                 # combine into one fraction
print(sp.apart((2 * x + 1) / (x * (x + 1))))            # split back into partial fractions
E_, U_, P_ = sp.symbols("E U P", positive=True)
identity = (1 - (E_ / P_) / ((E_ + U_) / P_)) - U_ / (E_ + U_)
print("Identity check (should be 0):", sp.simplify(identity))
````

## ⚠️ Common mistakes
- **Cancelling terms instead of factors:** $\dfrac{x + 3}{x + 5} \neq \dfrac{3}{5}$. You may only cancel a factor that multiplies the **whole** numerator and the **whole** denominator.
- Forgetting the domain restrictions after simplifying. $\frac{x^2-9}{x+3} = x - 3$ only for $x \neq -3$.
- Adding fractions by adding tops and bottoms: $\frac{1}{2} + \frac{1}{3} \neq \frac{2}{5}$.
- Dividing fractions without inverting the **second** fraction.
- Trying to factor $x^2 + 9$ over the reals.
- Mixing up the denominators of the rates: UR uses the **labor force**, LFPR uses the **population 15+**.
- Letting `inf` into your data by dividing by zero hours.

## Lesson 7.1 · Linear equations (including fractions) ⏱️ 20 min

### 📖 Step 1 · Read & Understand: linear equations

> **Concept card**
>
> A **linear equation** in $x$ can be written $ax + b = 0$ with $a \ne 0$, and it has exactly one solution $x = -b/a$.
> **Properties of equality:** you may add or subtract the same quantity on both sides, and multiply or divide both sides by the same **nonzero** number.
> **Fractions:** multiply both sides by the **LCD** to clear the denominators. Check that no denominator becomes 0.
> **Always check** by substituting back into the *original* equation.
>
> 📚 *Read more:* Swokowski & Cole, *Equations* (linear equations); Rees, Sparks & Rees, *Linear Equations*; Sydsaeter et al., *Solving Equations*.

### ✍️ Step 2 · Solve by Hand
**(a)** At ₱62.50/hour, how many hours give ₱3,000 per week?
$$62.5h = 3000 \;\Rightarrow\; h = \frac{3000}{62.5} = 48 \quad \text{(divide both sides by } 62.5 \ne 0).$$
*Check:* $62.5 \times 48 = 3000$ ✔️.

**(b)** After $\tfrac15$ of weekly pay $W$ goes to transport, ₱2,000 remains:
$$W - \tfrac{1}{5}W = 2000 \;\Rightarrow\; \tfrac45 W = 2000 \;\Rightarrow\; W = 2000 \cdot \tfrac54 = 2500.$$
**(c)** $\dfrac{h}{8} + \dfrac{h}{10} = 9$. Multiply both sides by the LCD 40: $5h + 4h = 360 \Rightarrow 9h = 360 \Rightarrow h = 40$.
### 🐍 Step 3 · Check with Python
New Python idea: `sp.Eq(left, right)` builds an equation, and `sp.solve(equation, variable)` solves it. The answer is a **list** of solutions.

````python
h, W_ = sp.symbols("h W")
print(sp.solve(sp.Eq(sp.Rational("62.5") * h, 3000), h))          # [48]
print(sp.solve(sp.Eq(W_ - W_ / 5, 2000), W_))                     # [2500]
print(sp.solve(sp.Eq(h / 8 + h / 10, 9), h))                       # [40]
print("Check (c):", sp.Rational(40, 8) + sp.Rational(40, 10))       # 9 ✔
````

## Lesson 7.2 · Quadratic equations and the peak of a profile ⏱️ 25 min

### 📖 Step 1 · Read & Understand: quadratic equations

> **Concept card**
>
> $ax^2 + bx + c = 0$ ($a \ne 0$) can be solved by:
> 1. **Factoring** and the zero-product property: $AB = 0 \Rightarrow A = 0$ or $B = 0$.
> 2. **Completing the square:** $x^2 + bx = \left(x + \tfrac{b}{2}\right)^2 - \tfrac{b^2}{4}$.
> 3. **Quadratic formula:** $x = \dfrac{-b \pm \sqrt{b^2 - 4ac}}{2a}$. The **discriminant** $D = b^2 - 4ac$ gives 2 real roots ($D > 0$), 1 ($D = 0$), or none ($D < 0$).
> **Vertex form:** $ax^2 + bx + c = a\left(x + \tfrac{b}{2a}\right)^2 + \left(c - \tfrac{b^2}{4a}\right)$. The vertex is at $x^* = -\tfrac{b}{2a}$. If $a < 0$ the parabola opens **down** and the vertex is a **maximum**.
>
> 📚 *Read more:* Swokowski & Cole, *Quadratic Equations*; Rees, Sparks & Rees, *Quadratic Equations*; Sydsaeter et al., *Solving Equations* (quadratic equations).

### ✍️ Step 2 · Solve by Hand
- **Factoring:** $x^2 - 6x + 5 = 0 \Rightarrow (x - 1)(x - 5) = 0 \Rightarrow x = 1$ or $x = 5$.
- **Completing the square:** $x^2 + 8x + 7 = 0 \Rightarrow x^2 + 8x + 16 = 9 \Rightarrow (x + 4)^2 = 9 \Rightarrow x + 4 = \pm 3 \Rightarrow x = -1$ or $x = -7$.
- **Formula:** $2x^2 - 3x - 1 = 0$: $D = 9 + 8 = 17$, so $x = \dfrac{3 \pm \sqrt{17}}{4}$.
- **Peak of a profile:** $p(a) = -0.2a^2 + 18a + 150$. Completing the square:
$$-0.2(a^2 - 90a) + 150 = -0.2\big[(a - 45)^2 - 2025\big] + 150 = -0.2(a - 45)^2 + 555.$$
Since $-0.2(a-45)^2 \le 0$, pay is **largest at $a^* = 45$**, where $p = 555$.
### 🐍 Step 3 · Check with Python

````python
x = sp.symbols("x")
print(sp.solve(x**2 - 6 * x + 5, x), sp.solve(x**2 + 8 * x + 7, x), sp.solve(2 * x**2 - 3 * x - 1, x))
age_s = sp.symbols("a")
demo = -sp.Rational(1, 5) * age_s**2 + 18 * age_s + 150
vertex_form = -sp.Rational(1, 5) * (age_s - 45)**2 + 555
print("Vertex form matches?", sp.expand(vertex_form - demo) == 0, "| peak age:", sp.Rational(-18) / (2 * -sp.Rational(1, 5)))
````

### 📊 Step 4 · Apply to LFS Data
**Question 1:** at what age does the fitted profile $p(a) = c_0 + c_1 a + c_2 a^2$ peak? Using the vertex formula, $a^* = -c_1/(2c_2)$. This is a maximum only if $c_2 < 0$.
**Question 2:** at what ages is predicted pay **equal to the sample median**? Solve $p(a) = m$. We use `sp.solveset(..., domain=sp.S.Reals)` so that only **real** roots come back, and then keep only realistic ages (15–99).

````python
c2, c1, c0 = np.polyfit(paid["age"], paid["daily_pay"], 2)          # refit (same as Part 1)
if c2 < 0:
    peak_age = -c1 / (2 * c2)                                          # vertex: a* = −b/(2a)
    peak_pay = c0 - c1**2 / (4 * c2)                                   # vertex value: c − b²/(4a)
    print(f"The profile opens DOWN (c2 = {c2:.5f} < 0): peak at age {peak_age:.1f}, ₱{peak_pay:,.2f}/day")
else:
    print("c2 ≥ 0: the fitted profile opens UP, so it has no peak (a minimum instead).")
````

````python
a = sp.symbols("a", real=True)
p_expr = c0 + c1 * a + c2 * a**2
m = paid["daily_pay"].median()
roots = sp.solveset(sp.Eq(p_expr, m), a, domain=sp.S.Reals)          # real solutions only
print("All real roots:", roots)
realistic = [float(r) for r in roots if 15 <= float(r) <= 99]         # keep plausible ages (a short loop inside [ ])
print("Ages (15–99) where predicted pay equals the median:", [round(v, 1) for v in realistic])
````

> ℹ️ `[float(r) for r in roots if ...]` is a **list comprehension**, a compact one-line `for` loop that builds a list. It is optional syntax. The long version would be: `realistic = []`, then `for r in roots:` → `if 15 <= float(r) <= 99: realistic.append(float(r))`.

````python
grid = np.linspace(15, 80, 200)
fig, ax = plt.subplots()
ax.plot(grid, c0 + c1 * grid + c2 * grid**2, lw=3, label="Fitted profile p(a)")
ax.axhline(m, color=OKABE_ITO[3], ls="--", label=f"Sample median ₱{m:,.0f}")
for r_ in realistic:
    ax.plot(r_, m, "o", ms=10, color="black")                          # where p(a) = m
ax.set_title("Where does the fitted age–pay profile equal the median?")
ax.set_xlabel("Age (years)"); ax.set_ylabel("Predicted daily pay (₱)"); ax.legend(); plt.show()
````

*Alt text:* a hill-shaped curve and a horizontal dashed line at the median pay. Black dots mark where they cross.
**Takeaway:** a quadratic equation can have two solutions: one on the rising side of the hill and one on the falling side. Between them, predicted pay is **above** the median (an inequality, coming in Part 8).

## Lesson 7.3 · Radical and absolute-value equations; extraneous roots ⏱️ 20 min

### 📖 Step 1 · Read & Understand: radical and absolute-value equations

> **Concept card**
>
> **Radical equations:** isolate the radical, then raise both sides to a power. Squaring can create **extraneous roots**, which are solutions of the new equation but not of the original. **Check every candidate** in the original equation.
> **Absolute value:** for $k \ge 0$, $|X| = k \iff X = k$ or $X = -k$. If $k < 0$ there is no solution.
> **Growth rates:** $P(1 + r)^2 = Q \Rightarrow 1 + r = \pm\sqrt{Q/P}$. Only the **positive** root makes economic sense, because $1 + r > 0$.
>
> 📚 *Read more:* Swokowski & Cole, *Other Types of Equations* (radicals, absolute values, extraneous solutions); Sydsaeter et al., *Solving Equations*.

### ✍️ Step 2 · Solve by Hand
**(a)** $\sqrt{x + 6} = x$. Square both sides: $x + 6 = x^2 \Rightarrow x^2 - x - 6 = 0 \Rightarrow (x - 3)(x + 2) = 0 \Rightarrow x = 3$ or $x = -2$.
Check: $x = 3$: $\sqrt{9} = 3$ ✔️. $x = -2$: $\sqrt{4} = 2 \ne -2$ ✘ **extraneous**. Solution: $x = 3$.

**(b)** $|x - 500| = 120 \Rightarrow x - 500 = 120$ or $x - 500 = -120 \Rightarrow x = 620$ or $x = 380$: the two pay levels exactly ₱120 from ₱500.

**(c)** Pay grows from ₱400 to ₱484 in 2 years: $400(1+r)^2 = 484 \Rightarrow (1+r)^2 = 1.21 \Rightarrow 1 + r = \pm 1.1$. We reject $-1.1$, so $r = 0.10$ (10% per year).
### 🐍 Step 3 · Check with Python

````python
print("Squared equation roots:", sp.solve(x**2 - x - 6, x))          # [-2, 3] includes the extraneous root
print("Original radical equation:", sp.solve(sp.sqrt(x + 6) - x, x))  # SymPy checks the roots → [3]
for cand in [3, -2]:
    print(f"x = {cand}: √(x+6) = {sp.sqrt(cand + 6)}, x = {cand} → {'✔' if sp.sqrt(cand + 6) == cand else '✘ extraneous'}")
print("|x − 500| = 120:", sp.solveset(sp.Abs(x - 500) - 120, x, domain=sp.S.Reals))
````

## Lesson 8.1 · Linear, quadratic and absolute-value inequalities ⏱️ 25 min

### 📖 Step 1 · Read & Understand: solving inequalities

> **Concept card**
>
> - **Linear:** same steps as an equation; **flip** the sign when you multiply or divide by a negative.
> - **Quadratic** (and other factorable ones), using a **sign chart**: (1) move everything to one side; (2) factor; (3) mark the zeros on a number line; (4) test the sign in each region; (5) choose the regions that satisfy the inequality, with endpoints included for $\le$ / $\ge$.
> - **Absolute value** ($k > 0$): $|X| < k \iff -k < X < k$ (*within* $k$), and $|X| > k \iff X < -k$ or $X > k$ (*farther than* $k$).
>
> 📚 *Read more:* Swokowski & Cole, *Inequalities* (sign charts; inequalities with absolute values); Sydsaeter et al., *Algebra* (inequalities).

### ✍️ Step 2 · Solve by Hand

| Region | Test $x$ | $x - 7$ | $x + 2$ | Product |
|---|---|---|---|---|
| $x < -2$ | $-3$ | $-$ | $-$ | $+$ |
| $-2 < x < 7$ | $0$ | $-$ | $+$ | $-$ ✔️ |
| $x > 7$ | $8$ | $+$ | $+$ | $+$ |

Solution: $(-2, 7)$.
**(b)** $|h - 40| \le 8 \iff -8 \le h - 40 \le 8 \iff 32 \le h \le 48$: hours *within 8* of a 40-hour week, $[32, 48]$.

## Lesson 8.2 · Systems of two linear equations ⏱️ 25 min

### 📖 Step 1 · Read & Understand: 2×2 linear systems

> **Concept card**
>
> A **system** asks for values that satisfy **all** the equations at once. Graphically, it is the **intersection** of the lines.
> - **Substitution:** solve one equation for one variable and substitute it into the other.
> - **Elimination:** add multiples of the equations to cancel a variable.
> - Three possibilities: **one** solution (lines cross, *consistent*), **none** (parallel lines, *inconsistent*), **infinitely many** (the same line, *dependent*).
> *Economics:* market equilibrium solves $Q_d = Q_s$. For example, $Q = 100 - 2P$ and $Q = 20 + 2P$ give $P = 20$, $Q = 60$.
>
> 📚 *Read more:* Rees, Sparks & Rees, *Systems of Equations*; Sydsaeter et al., *Solving Equations* (two equations with two unknowns); Chiang & Wainwright, *Partial Market Equilibrium: A Linear Model*.

### ✍️ Step 2 · Solve by Hand
$$\begin{cases} x + y = 10 \\ 3x + 5y = 38 \end{cases}$$
**Substitution:** $x = 10 - y \Rightarrow 3(10 - y) + 5y = 38 \Rightarrow 30 + 2y = 38 \Rightarrow y = 4$, so $x = 6$.
**Elimination:** multiply the first equation by 3: $3x + 3y = 30$. Subtract it from the second: $2y = 8 \Rightarrow y = 4$, and then $x = 6$.
**Check:** $6 + 4 = 10$ ✔️, $18 + 20 = 38$ ✔️.
### 🐍 Step 3 · Check with Python
New Python idea: pass a **list** of equations and a list of unknowns to `sp.solve`. The answer is a dictionary.

````python
xs, ys = sp.symbols("x y")
print(sp.solve([sp.Eq(xs + ys, 10), sp.Eq(3 * xs + 5 * ys, 38)], [xs, ys]))
Pp, Qq = sp.symbols("P Q")
print("Market equilibrium:", sp.solve([sp.Eq(Qq, 100 - 2 * Pp), sp.Eq(Qq, 20 + 2 * Pp)], [Pp, Qq]))
````

### 📊 Step 4 · Apply to LFS Data: recovering group sizes from totals
Suppose a report gives only the **total number** of paid workers $N$, their **total** daily pay $T$, and the **average** daily pay of college graduates ($\bar d_C$) and of everyone else ($\bar d_O$). How many workers are in each group? Let $x$ = college graduates and $y$ = others:
$$\begin{cases} x + y = N \\ \bar d_C\, x + \bar d_O\, y = T \end{cases}$$

````python
is_college = paid["educ"] == "College or higher"
N_total, T_total = len(paid), paid["daily_pay"].sum()
d_C = paid[is_college]["daily_pay"].mean()
d_O = paid[~is_college]["daily_pay"].mean()
sol = sp.solve([sp.Eq(xs + ys, N_total), sp.Eq(d_C * xs + d_O * ys, T_total)], [xs, ys])
print(f"Solved:  x (college) = {float(sol[xs]):.1f},  y (others) = {float(sol[ys]):.1f}")
print(f"Actual:  college = {is_college.sum()},  others = {(~is_college).sum()}")
````

````python
xgrid = np.linspace(0, N_total, 100)
fig, ax = plt.subplots()
ax.plot(xgrid, N_total - xgrid, lw=2.5, label="x + y = N  (headcount)")
ax.plot(xgrid, (T_total - d_C * xgrid) / d_O, lw=2.5, ls="--", label="d̄_C·x + d̄_O·y = T  (pay bill)")
ax.plot(float(sol[xs]), float(sol[ys]), "ko", ms=10, label="Intersection = solution")
ax.set_xlim(0, N_total); ax.set_ylim(0, N_total * 1.2)
ax.set_title("Two equations, two unknowns: recovering group sizes")
ax.set_xlabel("x = number of college graduates"); ax.set_ylabel("y = number of other paid workers"); ax.legend(); plt.show()
````

*Alt text:* two straight lines falling from left to right with different slopes, crossing at one black dot.
**Takeaway:** the system has **one** solution because the lines have different slopes ($-1$ versus $-\bar d_C/\bar d_O$), which happens exactly when the two group averages differ.

## Lesson 8.3 · Matrices, 3×3 systems, and types of systems ⏱️ 20 min

### 📖 Step 1 · Read & Understand: matrix form of a linear system

> **Concept card**
>
> A system can be written as $A\mathbf{x} = \mathbf{b}$: $A$ holds the coefficients, $\mathbf{x}$ the unknowns, and $\mathbf{b}$ the right-hand sides.
> For a 2×2 matrix $A = \begin{pmatrix} a & b \\ c & d \end{pmatrix}$, the **determinant** is $\det A = ad - bc$.
> - $\det A \ne 0$ ⇒ exactly one solution.
> - $\det A = 0$ ⇒ no solution (inconsistent) or infinitely many (dependent).
> For 3×3 systems, use elimination by hand (reduce to two equations in two unknowns) or a matrix solver to check.
>
> 📚 *Read more:* Rees, Sparks & Rees, *Systems of Equations* (determinants); Sydsaeter et al., *Solving Equations*; Chiang & Wainwright, *Linear Models and Matrix Algebra* (preview).

### ✍️ Step 2 · Solve by Hand
**(a)** $\det\begin{pmatrix} 1 & 1 \\ 3 & 5\end{pmatrix} = 5 - 3 = 2 \ne 0$, so the system in Lesson 8.2 has a unique solution.
**(b)** $x + y = 10,\; 2x + 2y = 30$. Doubling the first gives $2x + 2y = 20 \ne 30$, so it is **inconsistent** (parallel lines), with $\det = 2 - 2 = 0$.
**(c)** $x + y = 10,\; 2x + 2y = 20$ is **dependent**: every point with $y = 10 - x$ works.
**(d)** 3×3: $x + y + z = 6$, $x - y = 0$, $y + 2z = 7$. From the second, $x = y$. Then $2y + z = 6$ and $y + 2z = 7$. Eliminating gives $y = \tfrac53$, $z = \tfrac83$, $x = \tfrac53$.
### 🐍 Step 3 · Check with Python
New Python idea: `np.array` builds a matrix, `np.linalg.det` computes its determinant, and `np.linalg.solve(A, b)` solves $A\mathbf{x} = \mathbf{b}$.

````python
A2 = np.array([[1, 1], [3, 5]])
print("det =", round(np.linalg.det(A2), 6), "| solution:", np.linalg.solve(A2, np.array([10, 38])))
print("Inconsistent:", sp.solve([sp.Eq(xs + ys, 10), sp.Eq(2 * xs + 2 * ys, 30)], [xs, ys]))   # [] = no solution
print("Dependent:   ", sp.solve([sp.Eq(xs + ys, 10), sp.Eq(2 * xs + 2 * ys, 20)], [xs, ys]))   # x = 10 − y
A3 = np.array([[1, 1, 1], [1, -1, 0], [0, 1, 2]])
print("3×3 solution:", np.linalg.solve(A3, np.array([6, 0, 7])), "| 5/3 ≈", round(5 / 3, 4), ", 8/3 ≈", round(8 / 3, 4))
````

## ⚠️ Common mistakes
- Solving $x^2 < 9$ as $x < \pm 3$. ✅ Correct: $-3 < x < 3$.
- Reading the sign-chart regions backwards (using $>$ where the problem asks for $<$).
- Writing $|X| > k$ as one compound inequality $-k > X > k$, which is impossible. It is an **"or"**: $X < -k$ **or** $X > k$.
- Arithmetic slips in elimination: multiply **every** term of the equation.
- Concluding "no solution" when the system is actually **dependent** (infinitely many solutions).
- Calling `np.linalg.solve` on a singular matrix ($\det = 0$) gives an error. Check the determinant or use `sp.solve`.

## Lesson 9.2 · Writing a clean function ⏱️ 15 min

### 📖 Step 1 · Read & Understand: functions as reusable models
> A good function has a **clear name**, a **docstring** (the text in triple quotes that says what it does), **inputs**, and a **return** value. It also handles impossible cases (like dividing by zero) explicitly. It is the programming version of defining $f(x)$ *together with its domain*.

### ✍️ Step 2 · Solve by Hand
For a group with $E = 90$, $U = 10$, $P = 150$: LFPR $= \frac{100}{150} = 0.\overline{6}$ and UR $= \frac{10}{100} = 0.1$. If $E + U = 0$, the UR is **undefined** (division by zero).
### 🐍 Step 3 · Check with Python

````python
def labor_rates(E, U, P):
    """Return (LFPR, UR) from counts. UR is NaN when the labor force E + U is 0 (undefined)."""
    LF = E + U
    lfpr = LF / P
    if LF > 0:
        ur = U / LF
    else:
        ur = np.nan                    # domain restriction: denominator must be nonzero
    return lfpr, ur

print(labor_rates(90, 10, 150))       # (0.666..., 0.1)
print(labor_rates(0, 0, 20))          # (0.0, nan): no division-by-zero crash
````

````python
def labor_profile(data):
    """Teaching indicators (unweighted) for a table of persons aged 15+."""
    E = (data["empstat"] == "Employed").sum()
    U = (data["empstat"] == "Unemployed").sum()
    lfpr, ur = labor_rates(E, U, len(data))
    with_pay = data.dropna(subset=["daily_pay"])
    return {"persons": len(data), "LFPR": lfpr, "UR": ur,
            "median_daily_pay": with_pay["daily_pay"].median(),
            "median_hourly_pay": with_pay["hourly_pay"].median()}

summary = pd.DataFrame({"All": labor_profile(df), "NCR": labor_profile(df[df["region"] == "NCR"]),
                        "Women": labor_profile(df[df["sex"] == "Female"]),
                        "Youth 15–24": labor_profile(df[df["age"].between(15, 24)])})
summary.round(3)
````

````python
for sex in ["Female", "Male"]:
    g = df[df["sex"] == sex]
    E_g, U_g, P_g = (g["empstat"] == "Employed").sum(), (g["empstat"] == "Unemployed").sum(), len(g)
    lfpr_g, ur_g = labor_rates(E_g, U_g, P_g)
    print(f"Q3 {sex:<6}: UR = {ur_g:.4f}, 1 − EPR/LFPR = {1 - (E_g / P_g) / lfpr_g:.4f}")
````

### Q4 · Exponents and radicals (Week 3): at what constant yearly growth rate would the youth median daily pay reach today's adult median in 5 years?
*Setup:* $m_Y (1 + r)^5 = m_A \Rightarrow r = (m_A/m_Y)^{1/5} - 1$.

````python
m_Y = youth["daily_pay"].median()
m_A = adults["daily_pay"].median()
r_catch = (m_A / m_Y) ** (1 / 5) - 1
print(f"Q4: youth median ₱{m_Y:,.2f}, adult median ₱{m_A:,.2f} → r = {r_catch:.2%} per year")
````

### Q5 · Equations (Week 6): (a) at what age does the fitted profile peak? (b) How many weekly hours at the median hourly pay give ₱3,500?
*Setup:* (a) vertex $a^* = -c_1/(2c_2)$; (b) linear equation $r h = 3500$.

````python
c2, c1, c0 = np.polyfit(paid["age"], paid["daily_pay"], 2)
peak = -c1 / (2 * c2)
h_35 = 3500 / paid["hourly_pay"].median()
print(f"Q5a: profile opens {'down' if c2 < 0 else 'UP (no peak!)'}; vertex at age {peak:.1f}")
print(f"Q5b: hours needed for ₱3,500/week at the median hourly pay = {h_35:.1f}")
````

### Q6 · Inequalities (Week 7): (a) for which ages is predicted pay at least the median? (b) What share of employed persons work 48 hours or more, $h \ge 48$ (long hours)?

````python
a = sp.symbols("a", real=True)
m = paid["daily_pay"].median()
q6a = sp.solveset(c0 + c1 * a + c2 * a**2 >= m, a, domain=sp.S.Reals).intersect(sp.Interval(15, 99))
q6b = (has_hours["hours_week"] >= 48).mean()
print("Q6a: ages with predicted pay ≥ median:", q6a)
print(f"Q6b: share of employed (with hours) working h ≥ 48: {q6b:.1%}")
````

### Q7 · A 2×2 system (Week 7): recover the number of **youth** and **adult** paid workers from totals and averages alone.
*Setup:* $x + y = N$ and $\bar d_Y x + \bar d_A y = T$.

````python
paid_youth = paid["age"].between(15, 24)
d_Y, d_A = paid[paid_youth]["daily_pay"].mean(), paid[~paid_youth]["daily_pay"].mean()
xq, yq = sp.symbols("x_youth y_adult")
q7 = sp.solve([sp.Eq(xq + yq, len(paid)), sp.Eq(d_Y * xq + d_A * yq, paid["daily_pay"].sum())], [xq, yq])
print(f"Q7: solved youth = {float(q7[xq]):.0f}, adults = {float(q7[yq]):.0f}  | actual: {paid_youth.sum()}, {(~paid_youth).sum()}")
````
